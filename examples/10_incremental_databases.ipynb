# 10 — Incremental databases

**Audience:** Users attributing changes in LCIA results to ordered groups of Premise sector transformations.

**Prerequisites:** A configured Brightway project, a valid IAM key, and familiarity with superstructure analysis in Activity Browser.

**Learning goals:** define ordered update steps, build cumulative scenarios, and export an incremental superstructure.


## Outline

1. Configure an `IncrementalDatabase`.
2. Define ordered sector steps and groupings.
3. Export the cumulative scenarios.


```python
import os

import bw2data as bd

from premise import IncrementalDatabase

PROJECT = "ecoinvent-3.12-cutoff"
SOURCE_DATABASE = "ecoinvent-3.12-cutoff"
BIOSPHERE_DATABASE = "ecoinvent-3.12-biosphere"
PREMISE_KEY = os.environ.get("PREMISE_KEY")

if not PREMISE_KEY:
    raise RuntimeError("Set PREMISE_KEY before running this tutorial.")
bd.projects.set_current(PROJECT)
```

## 1. Create the incremental builder


```python
ndb = IncrementalDatabase(
    scenarios=[
        {"model": "image", "pathway": "SSP2-M", "year": 2050},
    ],
    source_db=SOURCE_DATABASE,
    source_version="3.12",
    biosphere_name=BIOSPHERE_DATABASE,
    key=PREMISE_KEY,
)
```

## 2. Define cumulative steps

Dictionary insertion order is the update order. A list groups several transformations into one attribution step.


```python
SECTOR_STEPS = {
    "electricity": "electricity",
    "steel": "steel",
    "transport and fuels": ["cement", "cars", "fuels"],
}
ndb.update(sectors=SECTOR_STEPS)
```

## 3. Export the incremental superstructure

Load the resulting scenario-difference file with the matching union database in Activity Browser.

![Incremental database analysis in Activity Browser](assets/incremental_database.png)


```python
ndb.write_increment_db_to_brightway(
    name="premise-image-2050-incremental",
    file_format="csv",
)
```

## Pitfalls and extension

- Incremental results depend on update order because later steps build on earlier ones.
- Group only sectors whose combined contribution answers the research question.
- Document the sector dictionary alongside reported contribution charts.
